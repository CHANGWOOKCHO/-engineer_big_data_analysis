## S11.슬라이싱, 사분위수, 결측치 제거

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

#### 1.데이터에서 결측치가 있는 데이터(행)을 모두 제거하시오.

```python
df = df.dropna()
print("행 수 :", len(df))
```

#### 2.결측치가 제거된 데이터를 사용하여 앞에서부터 70% 데이터를 구하시오.
#### (단, 데이터 70 % 지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)

```python
end = int(len(df) * 0.7)
df = df.iloc[:end]

print("행 수 :", len(df))
df.head()
```

#### 앞에서 구한 70% 데이터 중 'views' 컬럼의 3사분위수에서 1사분위수를 뺀 값을 소수점 이하를 버리고 정수부분만 구하시오.

```python
q3 = df['views'].quantile(0.75)
q1 = df['views'].quantile(0.25)

iqr = int(q3-q1)
print("IQR :", iqr)
```

## S12.결측치 처리, 최빈갑스 데이터 개수

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.결측치가 가장 많은 두 컬럼을 찾으시오.

```python
null_series = df.isnull().sum().sort_values(ascending=False)
null_df = null_series.to_frame(name='null_cnt')

col_nm1 = null_df.index[0]
col_nm2 = null_df.index[1]

print(f'1번째 결측가 많은 컬럼(결측치 수) : {null_df.index[0]}({null_df.iloc[0,0]})')
print(f'2번째 결측가 많은 컬럼(결측치 수) : {null_df.index[1]}({null_df.iloc[1,0]})')
```

#### 2.첫 번째로 결측치가 많은 컬럼에서 결측치가 있는 데이터(행)를 삭제하시오.

```python
df = df.dropna(subset=[col_nm1])

df.isnull().sum()
```

#### 3.두 번째로 결측치가 많은 컬럼을 최빈값으로 대체하시오.

```python
mode_value = df[col_nm2].mode()[0]
df[col_nm2] = df[col_nm2].fillna(mode_value)

df.isnull().sum()
```

#### 4.'f3'컬럼의 'gold'값을 가진 데이터의 수를 정수형으로 구하시오.

```python
cond = df['f3'] == 'gold'

print("gold 데이터의 수 :", sum(cond))
```
