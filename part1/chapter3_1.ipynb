## S11.슬라이싱, 사분위수, 결측치 제거

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

#### 1.데이터에서 결측치가 있는 데이터(행)을 모두 제거하시오.

```python
df = df.dropna()
print("행 수 :", len(df))
```

#### 2.결측치가 제거된 데이터를 사용하여 앞에서부터 70% 데이터를 구하시오.
#### (단, 데이터 70 % 지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)

```python
end = int(len(df) * 0.7)
df = df.iloc[:end]

print("행 수 :", len(df))
df.head()
```

#### 앞에서 구한 70% 데이터 중 'views' 컬럼의 3사분위수에서 1사분위수를 뺀 값을 소수점 이하를 버리고 정수부분만 구하시오.

```python
q3 = df['views'].quantile(0.75)
q1 = df['views'].quantile(0.25)

iqr = int(q3-q1)
print("IQR :", iqr)
```

## S12.결측치 처리, 최빈값 데이터 개수

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.결측치가 가장 많은 두 컬럼을 찾으시오.

```python
null_series = df.isnull().sum().sort_values(ascending=False)
null_df = null_series.to_frame(name='null_cnt')

col_nm1 = null_df.index[0]
col_nm2 = null_df.index[1]

print(f'1번째 결측가 많은 컬럼(결측치 수) : {null_df.index[0]}({null_df.iloc[0,0]})')
print(f'2번째 결측가 많은 컬럼(결측치 수) : {null_df.index[1]}({null_df.iloc[1,0]})')
```

#### 2.첫 번째로 결측치가 많은 컬럼에서 결측치가 있는 데이터(행)를 삭제하시오.

```python
df = df.dropna(subset=[col_nm1])

df.isnull().sum()
```

#### 3.두 번째로 결측치가 많은 컬럼을 최빈값으로 대체하시오.

```python
mode_value = df[col_nm2].mode()[0]
df[col_nm2] = df[col_nm2].fillna(mode_value)

df.isnull().sum()
```

#### 4.'f3'컬럼의 'gold'값을 가진 데이터의 수를 정수형으로 구하시오.

```python
cond = df['f3'] == 'gold'

print("gold 데이터의 수 :", sum(cond))
```

## S14.중복 데이터 제거, 값 변경, 데이터 개수

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.중복 데이터를 제거하시오.

```python
df= df.drop_duplicates()
print("행 수 :", len(df))
```

#### 2.'f3'컬럼의 결측치를 0, 'silver'는 1, 'gold'는 2, 'vip'는 3으로 변환하시오.

```python
mapping = {'silver' : 1, 'gold' : 2, 'vip' : 3}

df['f3'] = df['f3'].fillna(0)
df['f3'] = df['f3'].map(mapping)

df.head()
```

#### 3.변환된 'f3'컬럼의 총합을 정수형으로 구하시오.

```python
sum_value = int(df['f3'].sum())
print('f3의 총합 :', sum_value)
```

## S16.이상치, IQR

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.'views'컬럼의 1사분위수, 3사분위수 그리고 IQR을 계산하시오.

```python
q1 = df['views'].quantile(0.25)
q3 = df['views'].quantile(0.75)
iqr = q3 - q1

print("Q1 :", q1)
print("Q3 :", q3)
print("IQR : ", iqr)
```

#### 2.이상치 조건에 맞는 데이터를 찾으시오.
#### 이상치는 1사분위수 - (IQR * 1.5)보다 작은 값, 3사분위수 + (IQR * 1.5)보다 큰 값

```python
cond1 = df['views'] < q1 - (iqr * 1.5)
cond2 = df['views'] > q3 + (iqr * 1.5)
df = df[cond1 | cond2]

print("행 수 :", len(df))
df.head()
```

#### 3.이상치 데이터의 'views' 컬럼의 합을 정수로 구하시오.

```python
sum_value = int(df['views'].sum())
print("이상치의 합 :", sum_value)
```

## S17. 이상치, 소수점 있는 데이터 찾기, 표준편차 

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.'views' 컬럼의 표준편차를 구하시오.

```python
std_value1 = df['views'].std()
print("표준편차 :", std_value)
```

#### 2.'age'컬럼의 이상치(소수점 나이, 음수 나이, 0살)을 제거하고 'views' 컬럼의 표준편차를 구하시오.

```python
cond = df['age'] <= 0
df = df[~cond]

cond = df['age'] == round(df['age'],0)
df = df[cond]
```

#### 3.이상치 제거 전후의 'view'컬럼의 표준편차를 더하여, 반 올림 후 소수 둘째 자리까지 구하시오.

```python
std_value2 = df['views'].std()
sum_value = std_value1 + std_value2

print("표준편차의 합 :", round(sum_value, 2))
```

## S18.데이터(행) 기준 평균값, 인덱싱 

```python
df = pd.read_csv('type1_data2.csv', index_col='year')
print("행 수 :", len(df))
df.head()
```

#### 1.index '2001' 데이터(행)에서 평균보다 큰 값의 개수를 구하시오.

```python
mean2001 = df.loc[2001].mean()
cond1 = df.loc[2001] > mean2001

print("2001 데이터 평균 초과의 개수 :", sum(cond1))
```

#### 2.index '2003' 데이터(행)에서 평균보다 작은 값의 개수를 구하시오.

```python
mean2003 = df.loc[2003].mean()
cond2 = df.loc[2003] < mean2003

print("2001 데이터 평균 미만의 개수 :", sum(cond2))
```

#### 두 개수를 더하시오.

```python
print('합계 :', sum(cond1) + sum(cond2))
```

## S19.결측치(뒤의 값으로 대체), 그룹합

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.결측치를 바로 뒤에 있는 값으로 대체하시오.
#### (바로 뒤의 값도 결측치인 경우, 뒤에 있는 데이터 중 가장 가까운 값으로 대체)

```python
df = df.bfill()
df.head()
```

#### 2.'city'와 'f2' 컬럼을 기준으로 그룹합을 계산하시오.

```python
df = df.groupby(['city','f2']).sum(numeric_only=True).reset_index()
df
```

#### 3.'views'값이 세 번째로 큰 city 이름을 구하시오.

```python
df = df.sort_values('views', ascending=False)
df.head()
```

```python
print('3번째로 큰 도시명 :', df.iloc[2,0])
```

## S20.시계열 데이터, 월별 집계, 인덱스

```python
df = pd.read_csv('type1_data1.csv')
print("행 수 :", len(df))
df.head()
```

#### 1.연도 구분없이 월별로 숫자형 컬럼의 합을 구하시오.

```python
df['subscribed'] = pd.to_datetime(df['subscribed'], format='%Y-%m-%d')
df['month'] = df['subscribed'].dt.month
df.head()
```

```python
df = df.groupby(['month']).sum(numeric_only=True)
df.head()
```

#### 합계 중 'views'가 가장 작은 값을 가진 월을 정수로 구하시오.

```python
df = df.sort_values('views', ascending=True)
df.head()
```

```python
print("views가 가장 작은 달 : ", df.index[0])
```
