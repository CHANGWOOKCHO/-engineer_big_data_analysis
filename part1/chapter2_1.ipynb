```python
import pandas as pd
import numpy as numpy
```

```python
df = pd.DataFrame({
    "A" : ["데이터 분석", "기본 학습서", "퇴근 후 열공"],
    "B" : [10,20,30],
    "C" : ["ab cd", "AB CD", "ab cd"]
})

df
```

```python
#문자열 변경 -  replace로만은 전체문자열 중 일부 변경은 불가
df["A"] = df["A"].replace("분석", "시각화")
df
```

```python
df["A"] = df["A"].str.replace("분석", "시각화")
df
```

```python
#수치형 변경
df['B'] = df['B'].replace(10,100)
df
```

```python
#str.split()[0] vs str.split().str[0]
df['A'].str.split()[0]
```

```python
df['D'] = df['A'].str.split().str[0]
df
```

```python
#특정 문자열 탐색
df['A'].str.contains('기본')
```

```python
df['기본포함여부'] = df['A'].str.contains('기본')
df
```

```python
df['문자길이'] = df['A'].str.len()
df
```

```python
#소문자 변환
df['C'] = df['C'].str.lower()
df['C']
```

```python
df['C'] = df['C'].str.upper()
df['C']
```

```python
#형변환 + 문자변환
df['C'] = df['C'].str.lower()
df['C'] = df['C'].str.replace(' ', '')
df[['C']]
```

## 내장함수

```python
df = pd.read_csv('cafe4.csv')
df
```

```python
#len()
print("행 수 : ", str(len(df)))
print("행 수 : ", str(df.shape[0]))
print("열 수 : ", str(df.shape[1]))
```

```python
#조건에 맞는 행 수
cond = df['가격'] > 5000

#sum()으로 행 수
print("sum  : ", sum(cond))
#len()으로 행 수 
print("len : ", len(df[cond]))
```

```python
#데이터플레임의 sum(numeric_only=True) : 수치형 컬럼의 총합 반환
df.sum(numeric_only = True)
```

```python
df.describe()
```

```python
#기초 통계값
print("최댓값 :", df['가격'].max())
print("최솟값 :", df['가격'].min())
print("평균값 :", df['가격'].mean())
print("중앙값 :", df['가격'].median())
print("합계 :", df['가격'].sum())
print("표준편차 :", df['가격'].std())
print("분산 :", df['가격'].var())
```

```python
#quantile (4분위수)
```

```python
print('1분위수(25%) :',df['가격'].quantile(0.25))
print('중앙값(50%) :',df['가격'].quantile(0.5))
print('3분위수(75%) :',df['가격'].quantile(0.75))
```

```python
#1분위보다 낮은 값 찾기
cond = df['가격'].quantile(0.25) > df['가격']
df[cond]
```

```python
#3분위보다 큰 값 찾기
cond = df['가격'].quantile(0.75) < df['가격']
df[cond]
```
