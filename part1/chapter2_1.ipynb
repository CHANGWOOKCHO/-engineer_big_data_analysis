## melt() : 넓은 형태의 데이터를 긴 형태의 데이터로 변경
#### id_vars(필수) : 재구조 시 유지할 컬럼
#### value_vars(선택) : 재구조될 컬럼

```python
df = pd.DataFrame({
    '반' : ['A반','A반','B반'],
    '이름' : ['쿼카','알파카','시바견'],
    '수학' : [90,93,85],
    '영어' : [92,84,86],
    '국어' : [91,94,83],
    
})
df
```

```python
melt_df = pd.melt(df,id_vars=['이름'], value_vars=['수학','영어','국어'])
melt_df
```

```python
melt_df = pd.melt(df,id_vars=['반','이름'], value_vars=['국어', '영어'], var_name = '과목', value_name='점수')
melt_df
```

## transform()을 사용한 평균값 결측치 채우기
#### transform 통계함수 : mean(평균), sum(합계), 최댓값(max), 최솟값(min), 중앙값(median), 최빈값(mode)

```python
df = pd.DataFrame({
    '과일' : ['딸기', '블루베리','딸기', '블루베리','딸기', '블루베리','딸기', '블루베리'],
    '가격' : [1000, None, 1500, None, 2000, 2500, None, 1800],
})
df
```

```python
#transform()을 사용한 평균값 결측치 채우기
mean_price = df.groupby("과일")['가격'].transform("mean")
df['가격'] = df['가격'].fillna(mean_price)
df
```

```python
# 다중 그룹핑 조건
df = pd.DataFrame({
    '과일' : ['딸기', '블루베리','딸기', '블루베리','딸기', '블루베리','딸기', '블루베리'],
    '등급' : ['B','B','A','A', 'A', 'A', 'B', 'B'],
    '가격' : [1000, None, 1500, None, 2000, 2500, None, 1800],
})
df
```

```python
mean_price = df.groupby(['과일','등급'])['가격'].transform('mean')
mean_price
```

```python
df['가격'] = df['가격'].fillna(mean_price)
df
```

## unstack() : 다중인덱스에서 특정 인덱스를 컬럼으로 전환

```python
coffee_df = pd.DataFrame({
    '커피':['아메리카노','아메리카노','아메리카노','라떼','라떼','라떼'],
    '컵크기':['Small','Medium','Large','Small','Medium','Large'],
    '판매량':[120,150,200,100,130,180],
})
coffee_df
```

```python
grouped = coffee_df.groupby(['커피','컵크기'])['판매량'].sum()
grouped = grouped.to_frame()
```

```python
grouped1 = grouped.unstack() #level = -1(마지막 인덱스)가 기본값
grouped1
```

```python
grouped2 = grouped.unstack(level=0) #첫번째 인덱스를 컬럼으로 전환
grouped2
```
