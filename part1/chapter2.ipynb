## 정렬

```python
df = pd.read_csv("cafe2.csv")
df.head(2)
```

```python
#인덱스 기준
df.sort_index(ascending=False, inplace=True) #ascending : True(오름차순), False(내림차순)
df.head(3)
```

```python
# value 기준
df.sort_values("가격", ascending=False, inplace=True)
df.head(3)
```

```python
#다중 조건
df.sort_values(["가격","메뉴"], ascending=[False,True], inplace=True)
df.head()
```

```python
#인덱스 재설정
df.reset_index(drop=True, inplace=True)

df
```

## 필터링

```python
#not 연산
df[~cond]
```

```python
#주어진 값이 있는지로 조회
menus = ["녹차", "카푸치노","카페라떼"]
cond = df["메뉴"].isin(menus)

df[cond]
```
