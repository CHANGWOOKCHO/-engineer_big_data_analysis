```python
df = pd.read_csv("date_data.csv")
df['Date2'] = pd.to_datetime(df['Date2'], format='%Y:%m:%d')
df['Date3'] = pd.to_datetime(df['Date3'], format='%y/%m/%d')
df['DateTime1'] = pd.to_datetime(df['DateTime1'], format='%y-%m-%d %H:%M:%S')
df['DateTime2'] = pd.to_datetime(df['DateTime2'], format='%Y-%m-%d %H-%M-%S')
df[['Date2', 'Date3', 'DateTime1', 'DateTime2']]
```

```python
#날짜,시간 분해
df = df[['DateTime1']] #샘플 컬럼 1개만 남김
df = df.rename(columns={'DateTime1':'datetime'}) #컬럼명 변경
df
```

```python
df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['minute'] = df['datetime'].dt.minute
df['second'] = df['datetime'].dt.second
df['dayofweek'] = df['datetime'].dt.dayofweek
```

```python
# : map() 활용
week = {
    0:'일요일',
    1:'월요일',
    2:'화요일',
    3:'수요일',
    4:'목요일',
    5:'금요일',
    6:'토요일',
}

df['day_name'] = df['dayofweek'].map(week)
df
```

```python
#특정 시간과의 차이
day = pd.Timedelta(days=99)
df['100day'] = df['datetime']+ day
df
```

```python
#시간차이 구하기
diff = df['100day'] - df['datetime']
diff
```

```python
print('초 차이 :') 
print(diff.dt.total_seconds())
print('분 차이 :')
print(diff.dt.total_seconds()/60)
print('시간 차이 :')
print(diff.dt.total_seconds()/60/60)
print('일 차이 :')
print(diff.dt.total_seconds()/60/60/24)
```

## 데이터프레임 병합

```python
appetizer_df = pd.DataFrame({
    'Menu' : ['Salad', 'Soup', 'Bread'],
    'Price' : [5000,3000,2000]
})

main_df = pd.DataFrame({
    'Menu' : ['Steak','Pasta','Chicken'],
    'Price' : [15000,12000,10000]
})
```

```python
#concat 기본값은 행 병합
full_menu = pd.concat([appetizer_df,main_df], ignore_index=True)
full_menu
```

```python
#concat으로 열 병합
full_menu = pd.concat([appetizer_df, main_df], axis=1)
full_menu
```

```python
df = pd.DataFrame({
    "구분": ["전자", "전자", "전자", "전자", "전자", "가전", "가전", "가전", "가전"],
    "유형": ["일반", "일반", "일반", "특수", "특수", "일반", "일반", "특수", "특수"],
    "크기": ["소형", "대형", "대형", "소형", "소형", "대형", "소형", "소형", "대형"],
    "수량": [1, 2, 2, 3, 3, 4, 5, 6, 7],
    "금액": [2, 4, 5, 5, 6, 6, 8, 9, 9]
})
df
```

```python
pt = pd.pivot_table(
    df,
    index=['구분','유형'],
    values='수량',
    columns=['크기'],
    aggfunc='sum',
    fill_value=0
)
pt
```

```python
#여러 열에 대해 각기 다른 집계 함수 사용
pt = pd.pivot_table(
    df,
    values=['수량','금액'],
    index=['구분','크기'],
    aggfunc={'수량':'mean', '금액':['min','max','mean']}
)

pt
```

```python
pt=pt.reset_index()
pt
```
