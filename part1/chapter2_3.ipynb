```python
import numpy as np
import pandas as pd
```

## Q1 ~ Q10

```python
df = pd.DataFrame({
    '메뉴' : ['아메리카노','커페라떼','에스프레소','카페모카','바닐라라떼'],
    '가격' : [4500,5000,4000,5900,5300],
    '칼로리' : [10,110, np.nan, 210, np.nan],
    '원두' : ['과테말라', '브라질', '과테말라', np.nan, np.nan]
})
df
```

#### Q1.'칼로리' 컬럼의 결측치를 칼로리 데이터 중 최솟값으로 채우시오.

```python
min_col = df['칼로리'].min()
df['칼로리'] = df['칼로리'].fillna(min_col)
df
```

#### Q2.'원두' 컬럼의 결측치를 '원두'데이터 중 최빈값으로 채우시오.

```python
mode = df['원두'].mode()[0]
df['원두'] = df['원두'].fillna(mode)
df
```

#### Q3.가격이 5,000 이상인 데이터의 수를 구하시오.

```python
cond = df['가격'] >= 5000
print("sum 사용 :", sum(cond))
print("len 사용 :", len(df[cond]))
```

#### Q4.'이벤트가' 컬럼을 만들고 기존 가격에서 50% 할인된 가격을 채우시오.

```python
discount = 0.5
df['이벤트가'] = df['가격'] * (1-discount)
df['이벤트가'] = df['이벤트가'].astype('int')
df
```

#### Q5.'칼로리' 컬럼을 삭제하시오.

```python
df = df.drop('칼로리', axis=1)
df
```

#### Q6.위에서부터 3개의 행만 선택하시오.(loc사용)

```python
df.loc[:2]
```

#### Q7.위에서부터 3개의 행만 선택하시오.(iloc사용)
