```python
listbox = [15,46,78,24,56]
```

#### Q9.리스트에서 최댓값에서 최솟값을 뺀 결과를 리턴(반환)하는 함수를 만들고 실행하시오.

```python
def maxmin(list):
    return max(list) - min(list)
```

```python
print(maxmin(listbox))
```

# Q10

```python
str_data = "여러분의 합격을 응원합니다!"
```

#### Q10.문자열에서 "여러분"을 "당신"으로 변경한 결과를 리턴하는 함수를 만들고 실행하시오.

```python
def change_word(str):
    return str.replace("여러분", "당신")
```

```python
print(str_data)
print(change_word(str_data))
```
